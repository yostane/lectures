# Classes and types

Les classes permettent de modéliser les entités du métier.

Par exemple pour un meunuisier: Meuble (Furniture), Client, Order (commade)

```typescript
class Furniture {
  id;
  name;
  weight;
  price;

  constructor(id, name, weight, price){
    this.id = id;
    this.name = name;
    this.weight = weight;
    this.price = price;
  }
}

class Client {
  id;
  name;

  constructor(id, name){
    this.id = id;
    this.name = name;
  }
}

class Order {
  id;
  furnitures;
  client;
  // Si on suppose que les prix des meubles ne change pas. Le prix total peut être un getter
  get totalPrice() {
    return this.furnitures.reduce((acc, f) => acc + f.price, 0);
  }

  constructor(id, client){
    this.id = id;
    this.furnitures = [];
    this.client = client;
  }
}

const chair = new Furniture(1, "green chair", 10, 5);
const drawer = new Furniture(2, "Blue drawer", 20, 100);

const client = new Client(1, "Dupond");

const order = new Order(1, client);
order.furnitures.push(chair);
order.furnitures.push(chair, drawer);
console.log("Order", order.id, "total", order.totalPrice);
```
