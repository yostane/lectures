# Classes and types

Les classes permettent de modéliser les entités du métier.

Par exemple pour un meunuisier: Meuble (Furniture), Client, Order (commade)

```typescript
class Furniture {
  id;
  name;
  weight;
  price;

  constructor(id, name, weight, price){
    this.id = id;
    this.name = name;
    this.weight = weight;
    this.price = price;
  }
}

class Client {
  id;
  name;

  constructor(id, name){
    this.id = id;
    this.name = name;
  }
}

class Order {
  id;
  furnitures;
  client;

  constructor(id, client){
    this.id = id;
    this.furnitures = [];
    this.client = client;
  }
}
```
