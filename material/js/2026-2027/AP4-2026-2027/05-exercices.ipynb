## Personnages

```typescript
const baseUrl = "https://swapi.dev/api";
```

```typescript
// Typage d'une partie des propriétés
type People = {
  count: number;
  results: {
    name: string;
    height: string;
    eye_color: string;
  }[];
}

async function getPeople(): Promise<People>{
  const peopleResponse = await fetch(`${baseUrl}/people`);
  return await peopleResponse.json();
}
```

```typescript
// Exemples
const persons = await getPeople();
console.log("number of person in response", persons.count);
console.log("eye color of first person", persons.results[0].eye_color);
```

- Les noms des personnages de la première page

- La hauteur moyenne des personnages de la première page

- Le nombre de répétitions de chaque couleur d'yeux des personnages de la première page
