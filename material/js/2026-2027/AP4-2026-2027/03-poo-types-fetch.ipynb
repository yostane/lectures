# Classes and types

Les classes permettent de modéliser les entités du métier.

Par exemple pour un meunuisier: Meuble (Furniture), Client, Order (commade)

```typescript
class Furniture {
  id;
  name;
  weight;
  price;

  constructor(id, name, weight, price){
    this.id = id;
    this.name = name;
    this.weight = weight;
    this.price = price;
  }
}

class Client {
  id;
  name;

  constructor(id, name){
    this.id = id;
    this.name = name;
  }
}

class Order {
  id;
  furnitures;
  client;
  // Si on suppose que les prix des meubles ne change pas. Le prix total peut être un getter
  get totalPrice() {
    return this.furnitures.reduce((acc, f) => acc + f.price, 0);
  }

  constructor(id, client){
    this.id = id;
    this.furnitures = [];
    this.client = client;
  }
}

const chair = new Furniture(1, "green chair", 10, 5);
const drawer = new Furniture(2, "Blue drawer", 20, 100);

const client = new Client(1, "Dupond");

const order = new Order(1, client);
order.furnitures.push(chair);
order.furnitures.push(chair, drawer);
console.log("Order", order.id, "total", order.totalPrice);
```

Les objets littéraux sont définis directement sans classes

```typescript
const myLaptop = {
  name: "macbook",
  manufacturer: "Apple",
  apps: ["iMessage", "iMovie"]
}
console.log(myLaptop.name, myLaptop.manufacturer);
console.log(myLaptop.apps);
```

## La POO et TypeScript

- Readonly (comme const): on ne peut pas réassigner une propriété

```typescript
class Vehicle {
  constructor(readonly public id: number, public name:string) {}
}

const p206 = new Vehicle(1, "Peugeot 206");
console.log(p206);
```

```typescript
interface Ridable {
  readonly numberOfSeats: number;
}

class Berlin extends Vehicle implements Ridable {
  get numberOfSeats(){
    return 4;
  }
}

const v1 = new Berlin(2, "Peugeot 306");
console.log(v1, v1.numberOfSeats);
```

## Public et private

- Public: accessible depuis l'extérieur de la classe
- Private: accessible que depuis la classe
- protected: accessible depuis la classe et ses sous-classes

```typescript
class Animal {
  constructor(public name: string, private age: string) {}

  showAgeCategoty() {
    if (age < 5){
      console.log("baby");
    } else {
      console.log("adult");
    }
  }

  showAge(){
    console.log(this.age);
  }
}

const a = new Animal("Abou", 1);
console.log(a.name);
a.showAge()
```

## Fetch et async await

Fetch permet de télécharger des données. Elle retourne un promesse, donc il faut faire await fetch(....).
Il faut d'abord faite un await du fetch, puis une autre await du resultat.json()

```typescript
const response = await fetch("https://pokeapi.co/api/v2/pokemon/ditto")
const ditto = await response.json();
console.log(ditto.id, ditto.order);
console.log(ditto);
```
