# Premiers pas

JS est un langage dynamiquement (le type d'une variable peut changer) et implicitement typé (le langage déduit le type à partir de la valeur de la variable).

Conseil: décalrer **systématiquement** les variables avec `const` et changer en `let` si on a besoin de la réassigner.

```typescript
const x = "Hello";
console.log(x);
```

Types standard

```typescript
"Hello"; //string
12.23 // number
["Hello", "World"] // array
true; // booléen
null; // absence de valeur
undefined; // absence de valeur
NaN; // number
```

`undefined` est positionné par le langage ou par le code utilisateur ou des librairies.
`null` n'est jamais affecté par le langage mais unqiement par le code utilisateur ou des librairies.
