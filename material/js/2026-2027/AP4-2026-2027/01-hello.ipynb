## Fonctions et arguments undefined

```typescript
function add(a, b){
  console.log(a, b, a+b);
  return a + b;
}

add(1, 10, 100);
add(1, 10);
add(1);
add();

function multiply(a = 10, b = -6){
  console.log(a, b, a+b);
  return a * b;
}
```

Les fonctions sont des citoyens de première classe (en Anglais, on dit "1st class citizens"). C'est-à-dire que:

- On peut les affecter à des variables
- On peut les passer en argument de fonction
- Une fonction peut retourner une autre fonction

JS est compatible avec ces paradigmes dans une certaine mesure (il supporte la plupart des fonctionnalités mais pas toutes)
## Paradigme de programmation fonctionnelle

Basé sur l'exploitation des fonctionsà leur plein potentiel et l'immutabilité. Quelques caractéristiques:

- Fonction pures: 
  - Elle n'ont aucun effet sur l'environnement (elle ne modifient aucune variable ou état globaux)
  - Elle retourne toujours le même résultat pour les mêmes arguments en entrée
  - Exemple( add(x, y) -> pure)
- Fonctions sont des éléments de première classe
- Immutabilité -> que des constantes et si on souhaite, on créé une copie constante qui intègre les modifications

## Paradigme de programmation orientée objet

Style de programmation axé sur l'utilisation d'objet, classes et autres caratériques (encapsulation, héritage, polymorphisme, etc.)

## Programmation fonctionnelle et le JS

```typescript
function askQuestion(text){
  return text + "?";
}

const f = askQuestion;
console.log(askQuestion("How are you"));
console.log(f("Do you love JavaScript")); 
```

```typescript
// logBeforeAndAfter est qualifiée de fonction d'ordre supérieur car elle prend une fonction en argument
function logBeforeAndAfter(f){
  console.log("Running function");
  const result = f();
  console.log("After. Result is", result);
}

function sayHello(){
  return "Hello";
}
function sayBye(){
  return "Bye";
}

logBeforeAndAfter(sayHello);
logBeforeAndAfter(sayBye);
// Fonction anonyme, lambda
logBeforeAndAfter(function () {
  return "TypeScript is amazing!"
});
// Fonction flèche
logBeforeAndAfter(() => {
  return "TypeScript is awesome!"
});
```

En mode mutable. l'inconvénient et que l'interface risque d'être incohérente si on modifie une liste en cours d'utilisation

```typescript
const numbers = [2, -9, 4, 10];

// Ajouter 1 à chaque élément
for (let i = 0; i < numbers.length; i++) {
  numbers[i] += 1;
}

console.log(numbers)
```

Mode immutable avec [map](https://navin-moorthy.github.io/6ecb7bc629794f35fb63d4e0b92c05a5/map.gif), filter et reduce. ([illustrations](https://navin-moorthy.github.io/blog/map-filter-reduce-animated/))

```typescript
const numbers2 = [2, -9, 4, 10];
const resultNumbers = numbers2.map((number) => number + 1);

function increment(x){
  return x + 1;
}
console.log(numbers2.map(increment));

console.log("+1", resultNumbers);
// Récupérer les nomnbres pairs
const evenNumbers = numbers2.filter(x => x % 2 === 0);
console.log(evenNumbers);
```

Reduce permet de transformer une liste en une seule valeur.
- Somme
- Min et le max
- Nombre de lettres dans une phrase

```typescript
console.log("Somme des éléments", numbers2.reduce((acc, number) => acc + number));
console.log("Min", numbers2.reduce((acc, number) => acc < number ? acc : number, numbers2[0]));
```

```typescript
const words = ["i", "love", "JS"];
// nombre de lettres total des mots qui ont une voyelle
// Prédicat: fonction qui retourne true ou false
function isVowel(word){
  return word.includes("a") || word.includes("i") || word.includes("u") || word.includes("e") || word.includes("o")  
}

const result = words.filter(isVowel).map(word => word.length).reduce((acc, x)=> acc + x);
console.log(result);
```
